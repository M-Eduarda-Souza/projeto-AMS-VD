### Relatório de qualidade — TotalCharges

**Problema encontrado:** Foram identificados 11 valores ausentes ou não numéricos na variável `TotalCharges`, que representa o total de cobranças acumuladas por cliente.

**Decisão aplicada:** Em investigação. Os registros serão analisados em conjunto com o tempo de permanência (`tenure`) antes de definir o tratamento adequado.

**Impacto e validação:** A ocorrência afeta 11 registros da base. A quantidade corresponde a aproximadamente 0,16% das 7.043 observações da versão original da base. A decisão de tratamento será documentada e a quantidade de registros será conferida após o processamento.

```python
df = df_original.copy()

df.columns = df.columns.str.strip()

for coluna in df.select_dtypes(include="object").columns:
    df[coluna] = df[coluna].str.strip()

df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

df.loc[
    df["TotalCharges"].isna() & (df["tenure"] == 0),
    "TotalCharges"
] = 0


print("Dimensões da base:", df.shape)
print("Valores ausentes em TotalCharges:", df["TotalCharges"].isna().sum())
print("Linhas duplicadas:", df.duplicated().sum())
print("IDs duplicados:", df["customerID"].duplicated().sum())
```

### Variável: TotalCharges

**Problema encontrado:** Foram identificados 11 valores não numéricos ou ausentes em `TotalCharges`, todos associados a clientes com `tenure = 0`.

**Decisão aplicada:** Os 11 valores foram substituídos por zero na cópia processada da base, considerando que esses clientes não apresentam período de permanência registrado. O arquivo original foi preservado.

**Impacto e validação:** Foram mantidas as 11 observações, evitando perda de registros. A quantidade de valores ausentes foi verificada novamente após a transformação. A hipótese de cobrança acumulada igual a zero deverá ser considerada uma limitação da análise.

### Variável: Churn

**Problema encontrado:** A variável-alvo apresenta distribuição desigual entre as classes. Dos 7.043 clientes, 5.174 (73,46%) não cancelaram e 1.869 (26,54%) cancelaram. Não foram identificados valores ausentes.

**Decisão aplicada:** A variável será mantida como alvo de classificação binária, com as categorias “Yes” e “No”. A distribuição das classes será considerada na escolha das métricas e dos baselines.

**Impacto e validação:** A classe majoritária corresponde a 73,46% das observações. Uma previsão constante da classe majoritária teria essa acurácia na base completa, mas não identificaria nenhum cancelamento. O baseline deverá ser avaliado em um conjunto de validação separado.

```python
from pathlib import Path

pasta_processado = Path("dados/processado")
pasta_processado.mkdir(parents=True, exist_ok=True)

arquivo_saida = pasta_processado / "telco_churn_processado.csv"
df.to_csv(arquivo_saida, index=False)

print("Base processada salva em:", arquivo_saida.resolve())
print("Dimensões do arquivo salvo:", pd.read_csv(arquivo_saida).shape)
```

### Treino, Validação e Teste

Como a base não possui uma linha do tempo clara,e a separação será aleatória, mas precisamos garantir que a proporção de cancelamentos seja mantida.

```python
from sklearn.model_selection import train_test_split

X = df.drop(columns=['customerID', 'Churn'])
y = df['Churn'].map({'No': 0, 'Yes': 1}) 

X_temp, X_test, y_temp, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

X_train, X_val, y_train, y_val = train_test_split(
    X_temp, y_temp, test_size=0.25, random_state=42, stratify=y_temp
)

print("--- Dimensões dos Conjuntos ---")
print(f"Treino:    {X_train.shape[0]} linhas")
print(f"Validação: {X_val.shape[0]} linhas")
print(f"Teste:     {X_test.shape[0]} linhas")
```

### Pré-processamento sem vazamento (Pipeline)

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer


colunas_numericas = ['tenure', 'MonthlyCharges', 'TotalCharges']
colunas_categoricas = X.drop(columns=colunas_numericas).columns.tolist()

pipe_num = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

pipe_cat = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(drop='if_binary', handle_unknown='ignore', sparse_output=False))
])

pre_processador = ColumnTransformer(transformers=[
    ('num', pipe_num, colunas_numericas),
    ('cat', pipe_cat, colunas_categoricas)
])

X_train_processado = pre_processador.fit_transform(X_train)

X_val_processado = pre_processador.transform(X_val)
X_test_processado = pre_processador.transform(X_test)

print("--- Evidência de Transformação (Mesmo conjunto de atributos) ---")
print(f"Features no Treino:    {X_train_processado.shape[1]}")
print(f"Features na Validação: {X_val_processado.shape[1]}")
print(f"Features no Teste:     {X_test_processado.shape[1]}")
```

### Eficácia do Baseline Trivial e da Baseline Operacional

```python
from sklearn.metrics import accuracy_score, classification_report

y_pred_trivial = [0] * len(y_val)

acuracia_trivial = accuracy_score(y_val, y_pred_trivial)
print("--- BASELINE TRIVIAL ---")
print(f"Acurácia na Validação: {acuracia_trivial:.4f}")
print("Relatório de Classificação:")
print(classification_report(y_val, y_pred_trivial, zero_division=0))


def regra_operacional(row):
    if row['Contract'] == 'Month-to-month' and row['MonthlyCharges'] > 65.0:
        return 1
    return 0

y_pred_operacional = X_val.apply(regra_operacional, axis=1)

acuracia_operacional = accuracy_score(y_val, y_pred_operacional)
print("\n--- BASELINE OPERACIONAL ---")
print(f"Acurácia na Validação: {acuracia_operacional:.4f}")
print("Relatório de Classificação:")
print(classification_report(y_val, y_pred_operacional))
```
